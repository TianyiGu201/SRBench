```python
import os
import subprocess
import pandas as pd
import tifffile
from matplotlib import pyplot as plt
from PIL import Image
import numpy as np
import h5py
from scipy.io import mmread
from scipy.sparse import csr_matrix
```

```python
data_dir = "PATH/TO/DATASET/"
```

```python
cnts = pd.read_csv(
        f"{data_dir}/cnts.tsv",
        sep='\t',
        header=0
)
locs = pd.read_csv(
        f"{data_dir}/locs-raw.tsv",
        sep='\t',
        header=0
)
cnts.to_csv(f"{data_dir}cnts.csv", sep=',', index=False)
locs.to_csv(f"{data_dir}locs-raw.csv", sep=',', index=False)
```

```python
with open(f"{data_dir}radius-raw.txt","w") as f:
    f.write(str(round(55 / 2 / 1.7)))

with open(f"{data_dir}radius.txt","w") as f:
    f.write(str(int(55 / 2 / 1.7)))

with open(f"{data_dir}pixel-size-raw.txt","w") as f:
    f.write(str(1.7))
```
