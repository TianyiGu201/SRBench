# Load native 8-micrometre bin data

First export `cnts.tsv` and `locs-raw.tsv` from the native 8-micrometre matrix and matched positions with `export_native_bins.py` (see `docs/PREPROCESSING.md`). Do not use simulated coarse-resolution matrices for this step. To reproduce the study, retain the finalized study gene list instead of selecting it again.


```r
dataset_dir = "YOUR_PATH"
result_dir = "YOUR_PATH"
dir.create(result_dir, recursive = TRUE, showWarnings = FALSE)
```

```r
exp = read.csv(
    file.path(dataset_dir, "cnts.tsv"),
    sep = "\t", check.names=F
)
exp
```

```r
loc = read.csv(
    file.path(dataset_dir, "locs-raw.tsv"),
    sep = "\t", check.names=F
)
loc
```

```r
exp_df <- merge(loc, exp, by="spot")
rownames(exp_df) <- exp_df$spot
exp_df <- exp_df[,c(3, 2, 4:dim(exp_df)[2])]
colnames(exp_df)[c(1:2)] <- c("row", "col")
exp_df
```

# SVG calculate

```r
result <- heartsvg(exp_df,scale=T)
result
```

```r
result_sub = result[(result$rank <= 1000),]
print(result_sub)
SVGs <- result_sub$gene
```

```r
write.table(SVGs, paste0(result_dir, "/gene-names.txt"), quote = FALSE, row.names = FALSE, col.names = FALSE)
```
